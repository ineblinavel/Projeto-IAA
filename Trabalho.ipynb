```python
import json
import os

caminhos_json = [
    "/content/drive/MyDrive/yoloms_kfold_runs/yoloms_kfold_0/20250704_150923/vis_data/20250704_150923.json",
    "/content/drive/MyDrive/yoloms_kfold_runs/yoloms_kfold_1/20250704_153254/vis_data/20250704_153254.json",
    "/content/drive/MyDrive/yoloms_kfold_runs/yoloms_kfold_2/20250704_155813/vis_data/20250704_155813.json",
    "/content/drive/MyDrive/yoloms_kfold_runs/yoloms_kfold_3/20250704_161648/vis_data/20250704_161648.json",
    "/content/drive/MyDrive/yoloms_kfold_runs/yoloms_kfold_4/20250708_212004/vis_data/20250708_212004.json"
]

output_dir = "/content/runs_metrics"
os.makedirs(output_dir, exist_ok=True)

for idx, caminho in enumerate(caminhos_json):
    loss_logs = []
    metric_logs = []

    with open(caminho, "r") as f:
        lines = f.readlines()

    for line in lines:
        try:
            data = json.loads(line)
            if "coco/bbox_mAP" in data:
                metric_logs.append(data)
            elif "loss" in data:
                loss_logs.append(data)
        except json.JSONDecodeError:
            print(f"Linha inválida ignorada no arquivo {caminho}:", line)

    base_name = f"run{idx}"
    loss_path = os.path.join(output_dir, f"{base_name}_loss_logs.json")
    metric_path = os.path.join(output_dir, f"{base_name}_metric_logs.json")

    with open(loss_path, "w") as f_loss:
        json.dump(loss_logs, f_loss, indent=2)

    with open(metric_path, "w") as f_metric:
        json.dump(metric_logs, f_metric, indent=2)

    print(f"Arquivos salvos em {output_dir}:")
    print(f"- {base_name}_loss_logs.json")
    print(f"- {base_name}_metric_logs.json")
```

```python
import json
import os
import glob
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings

LOGS_DIR = '/content/runs_metrics/'
NUM_FOLDS = 5
LOSS_CHART_FILENAME = 'curva_de_perda_treinamento.png'
MAP50_CHART_FILENAME = 'curva_de_performance_map50.png'

warnings.simplefilter(action='ignore', category=FutureWarning)

all_train_losses = []
all_map50 = []

print("Coletando e processando os logs de todos os folds...")
print("-" * 50)

for k in range(NUM_FOLDS):
    print(f"Processando Fold {k}...")

    loss_file_path = os.path.join(LOGS_DIR, f'run{k}_loss_logs.json')
    if not os.path.exists(loss_file_path):
        print(f"  [AVISO] Arquivo de loss não encontrado: {loss_file_path}. Pulando.")
        continue

    loss_df = pd.read_json(loss_file_path)

    train_loss_per_epoch = loss_df.groupby('epoch')['loss'].mean().rename(f'fold_{k}')
    all_train_losses.append(train_loss_per_epoch)

    metric_file_path = os.path.join(LOGS_DIR, f'run{k}_metric_logs.json')
    if not os.path.exists(metric_file_path):
        print(f"  [AVISO] Arquivo de métricas não encontrado: {metric_file_path}. Pulando.")
        continue

    metric_df = pd.read_json(metric_file_path)
    metric_df['epoch'] = metric_df['step']

    map50_per_epoch = metric_df.set_index('epoch')['coco/bbox_mAP_50'].rename(f'fold_{k}')
    all_map50.append(map50_per_epoch)

    print(f"  Dados do Fold {k} carregados com sucesso.")

print("-" * 50)
print("Coleta de dados finalizada.")

if not all_map50 or not all_train_losses:
    print("\nERRO CRÍTICO: Não foi possível carregar os dados de loss ou métricas. Verifique os arquivos e caminhos.")
else:
    df_map50 = pd.concat(all_map50, axis=1)
    df_train_loss = pd.concat(all_train_losses, axis=1)

    mean_map50 = df_map50.mean(axis=1)
    std_map50 = df_map50.std(axis=1)

    mean_train_loss = df_train_loss.mean(axis=1)
    std_train_loss = df_train_loss.std(axis=1)

    epochs = mean_map50.index

    plt.style.use('seaborn-v0_8-whitegrid')

    fig1, ax1 = plt.subplots(figsize=(14, 7))
    ax1.plot(epochs, mean_train_loss, 'o-', color='royalblue', label='Perda Média de Treinamento')
    ax1.fill_between(epochs, mean_train_loss - std_train_loss, mean_train_loss + std_train_loss, color='royalblue', alpha=0.1, label='Desvio Padrão (Treino)')

    ax1.set_title('Curva de Perda de Treinamento (Média de 5 Folds)', fontsize=16, pad=20)
    ax1.set_xlabel('Época', fontsize=12)
    ax1.set_ylabel('Loss', fontsize=12)
    ax1.legend(loc='upper right', fontsize=12)
    ax1.set_xlim(left=1, right=epochs.max())
    ax1.set_ylim(bottom=0)

    plt.savefig(LOSS_CHART_FILENAME, dpi=300, bbox_inches='tight')
    plt.show()

    fig2, ax2 = plt.subplots(figsize=(14, 7))
    ax2.plot(epochs, mean_map50, 'o-', color='forestgreen', label='mAP@50 Médio na Validação')
    ax2.fill_between(epochs, mean_map50 - std_map50, mean_map50 + std_map50, color='forestgreen', alpha=0.1, label='Desvio Padrão')

    best_epoch_avg = mean_map50.idxmax()
    best_map50_avg = mean_map50.max()
    ax2.plot(best_epoch_avg, best_map50_avg, 'r*', markersize=15, label=f'Melhor Média: {best_map50_avg:.3f} (Época {best_epoch_avg})')

    ax2.set_title('Curva de Performance mAP@50 (Média de 5 Folds)', fontsize=16, pad=20)
    ax2.set_xlabel('Época', fontsize=12)
    ax2.set_ylabel('mAP@50', fontsize=12)
    ax2.legend(loc='lower right', fontsize=12)
    ax2.set_xlim(left=1, right=epochs.max())
    ax2.set_ylim(0, max(0.8, mean_map50.max() * 1.1))

    plt.savefig(MAP50_CHART_FILENAME, dpi=300, bbox_inches='tight')
    plt.show()

    best_scores_per_fold = df_map50.max()
    final_mean = best_scores_per_fold.mean()
    final_std = best_scores_per_fold.std()

    print("\n" + "="*50)
    print("--- RESUMO ESTATÍSTICO FINAL ---")
    print("="*50)
    print(f"\nGráfico da curva de perda salvo como: '{LOSS_CHART_FILENAME}'")
    print(f"Gráfico da curva de performance salvo como: '{MAP50_CHART_FILENAME}'")
    print("\nMelhor mAP@50 alcançado em cada fold:")
    print(best_scores_per_fold.round(4))
    print("\n" + "-"*50)
    print("--- RESULTADO PARA REPORTAR ---")
    print(f"Média dos melhores mAP@50: {final_mean:.4f}")
    print(f"Desvio padrão dos melhores mAP@50: {final_std:.4f}")
    print(f"\nAP₅₀ Final = {final_mean:.3f} ± {final_std:.3f}")
    print("="*50)
```
